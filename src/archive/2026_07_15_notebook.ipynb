# 1. GDELT

## Background

The **Global Database of Events, Language and Tone (GDELT)** is an open-source database that continuously monitors news media from around the world. It automatically extracts information from news articles and converts them into structured event records using the CAMEO event coding system. Each record contains details such as the event type, actors involved, geographic location, media attention and sentiment.

## Purpose

The primary purpose of GDELT is to provide a near real-time representation of political, social and economic events reported by global news sources. Unlike humanitarian datasets that rely on manually curated reports, GDELT captures events directly from news coverage.

## Information Provided

The dataset includes:

- Event codes (CAMEO)
- Event date
- Country and location
- Geographic coordinates
- Actor information
- Goldstein conflict score
- Tone of reporting
- Number of mentions, sources and articles

## Relevance to Food Insecurity

Although GDELT does not directly measure food insecurity, it captures many of the underlying drivers that may influence food security, including conflict, political instability, protests, violence and natural disasters. Its high temporal resolution makes it valuable for monitoring rapidly evolving situations.

---

# 2. ReliefWeb

## Background

**ReliefWeb** is a humanitarian information service managed by the United Nations Office for the Coordination of Humanitarian Affairs (UNOCHA). Rather than collecting individual events, ReliefWeb aggregates reports published by humanitarian organisations, governments, NGOs and UN agencies involved in disaster response and humanitarian assistance.

## Purpose

ReliefWeb aims to centralise humanitarian information and make reports accessible to governments, researchers and humanitarian practitioners. Reports cover a wide range of humanitarian issues, including food security, displacement, health, conflict and disaster response.

## Information Provided

The dataset contains:

- Report title
- Full report text
- Publication date
- Humanitarian themes
- Source organisation
- Report format
- Country coverage
- Original report URL

## Relevance to Food Insecurity

ReliefWeb provides detailed contextual information surrounding humanitarian crises. Unlike other datasets focused on aggregating news, it includes narrative descriptions that explain the causes, impacts and humanitarian response associated with food insecurity.

---

# 3. ACLED

## Background

The **Armed Conflict Location & Event Data (ACLED)** project is a conflict event database that records incidents of political violence, demonstrations and strategic developments. Events are manually coded using reports from news agencies, governments, NGOs and local organisations.

## Purpose

ACLED aims to provide detailed information on the occurrence, location and characteristics of conflict-related events worldwide.

## Information Provided

The dataset includes:

- Event type
- Sub-event type
- Event date
- Geographic coordinates
- Administrative regions
- Fatalities
- Actors involved
- Event description

## Relevance to Food Insecurity

Conflict is widely recognised as one of the major drivers of food insecurity. ACLED provides highly detailed conflict information that can be linked to changes in food availability, market disruption and population displacement.

---

# 4. ICEWS

## Background

The **Integrated Crisis Early Warning System (ICEWS)** is an event database developed to support conflict forecasting and political stability analysis. Events are extracted automatically from news reports using natural language processing techniques and coded according to the CAMEO framework.

## Purpose

ICEWS was developed to monitor political interactions and provide structured event data for crisis forecasting and early warning applications.

## Information Provided

The dataset contains:

- Event date
- Event code
- Source and target actors
- Geographic location
- Country
- Event description

## Relevance to Food Insecurity

ICEWS captures political interactions, protests, violence and other events that may contribute to humanitarian crises. Similar to GDELT, it provides structured event information suitable for temporal analysis.

---

# 5. CHIRPS

## Background

The **Climate Hazards Group InfraRed Precipitation with Station data (CHIRPS)** is a rainfall dataset produced by the Climate Hazards Center at the University of California, Santa Barbara. It combines satellite observations with ground-based weather station measurements to estimate rainfall.

## Purpose

CHIRPS was developed to monitor rainfall variability and support drought monitoring, agricultural planning and food security assessments.

## Information Provided

The dataset contains:

- Rainfall estimates
- Geographic coordinates
- Raster grid cells
- Monthly or daily observations
- Continuous spatial coverage

## Relevance to Food Insecurity

Rainfall is a key environmental driver of agricultural production and food availability. CHIRPS provides an objective measure of climatic conditions that may influence crop production and drought.

---

# 6. WFP Food Prices

## Background

The **World Food Programme (WFP) Vulnerability Analysis and Mapping (VAM)** programme collects food price information from local markets around the world. The data are compiled from national market monitoring systems and partner organisations.

## Purpose

The dataset is designed to monitor food prices and market conditions, providing evidence for food security assessments and humanitarian decision-making.

## Information Provided

The dataset includes:

- Commodity
- Market
- Administrative region
- Price
- Currency
- Standardised USD price
- Observation date

## Relevance to Food Insecurity

Food prices directly influence household purchasing power and access to food. Rising prices are commonly used as an indicator of economic access to food and are therefore an important component of food security monitoring.

# Section 4: Defining features

### Spatial and Temporal Unit of Analysis

The datasets included in this study have different structures and units of observation. GDELT, ACLED and ICEWS contain individual event records, ReliefWeb contains humanitarian reports, CHIRPS provides rainfall observations across raster grid cells, and WFP provides commodity price observations from individual markets. As a result, the datasets cannot be directly combined at the individual record level.

To allow comparison and integration across sources, a common spatial and temporal unit of analysis is required. For this study, the proposed unit of analysis is

                              **ADM2 × month**

where each row represents one second-level administrative area during a specific month.

The ADM2 level was selected because it provides a more detailed geographic unit than national or first-level administrative data. This allows local differences in conflict events, humanitarian reporting, rainfall and market conditions to be represented. Where geographic coordinates are available, records can be assigned to ADM2 areas using administrative boundaries. For datasets with different spatial structures, such as CHIRPS raster data and WFP market observations, the information can be aggregated to the corresponding ADM2 area.

A monthly temporal unit was selected to provide a common time interval across datasets with different reporting frequencies. High-frequency event datasets such as GDELT can be aggregated into monthly summaries, while monthly rainfall and market observations can be more easily aligned to the same period. The monthly level also reduces the effect of short-term variation while maintaining sufficient temporal detail to observe changes between periods.

- `single_source_only` – Indicates whether only one source provides information.

These features can be used to identify areas and periods where the information sources agree in their coverage and where important coverage gaps exist.

### 1. Spatial and Temporal Identifiers

These variables identify the geographic area and time period represented by each row.

- `country` – Country where the observation is located.
- `admin1` – First-level administrative area.
- `admin2` – Second-level administrative area.
- `year` – Year of observation.
- `month` – Month of observation.

### 2. Event / Record Count and Classification Features

These features describe the amount and type of information captured by GDELT, ACLED, ICEWS and ReliefWeb. Records from each source will be aggregated to the common **ADM2 × month** unit of analysis.

For each dataset, a total record count will first be calculated. The existing classification system provided by each source will then be used to calculate the number of records associated with each event type, event class or humanitarian theme.

This approach preserves the original classification structure of each dataset rather than applying a new researcher-defined classification.

#### GDELT

GDELT events are classified using the CAMEO event coding system. The `QuadClass` variable groups events into four broad categories.

- `gdelt_event_count` – Total number of GDELT events.
- `gdelt_verbal_cooperation_count` – Number of events classified as verbal cooperation.
- `gdelt_material_cooperation_count` – Number of events classified as material cooperation.
- `gdelt_verbal_conflict_count` – Number of events classified as verbal conflict.
- `gdelt_material_conflict_count` – Number of events classified as material conflict.

More detailed event categories may also be extracted using `EventRootCode` if required.

#### ACLED

ACLED records are classified using the event types provided by the dataset. The total number of events and the number of events within each classification will be calculated.

- `acled_event_count` – Total number of ACLED events.
- `acled_battle_count` – Number of events classified as Battles.
- `acled_violence_civilians_count` – Number of Violence against civilians events.
- `acled_protest_count` – Number of Protest events.
- `acled_riot_count` – Number of Riot events.
- `acled_explosion_remote_violence_count` – Number of Explosions/Remote violence events.
- `acled_strategic_development_count` – Number of Strategic developments.

The ACLED `sub_event_type` variable may later be used to provide a more detailed classification if required.

#### ICEWS

ICEWS also uses the CAMEO event coding framework. Events can therefore be grouped into the same four broad CAMEO event classes used for GDELT.

- `icews_event_count` – Total number of ICEWS events.
- `icews_verbal_cooperation_count` – Number of verbal cooperation events.
- `icews_material_cooperation_count` – Number of material cooperation events.
- `icews_verbal_conflict_count` – Number of verbal conflict events.
- `icews_material_conflict_count` – Number of material conflict events.

More detailed event classifications may also be extracted from the CAMEO event codes.

#### ReliefWeb

ReliefWeb differs from the event-based datasets because each record represents a humanitarian report rather than an individual event. However, the same aggregation approach can be applied by counting reports and using the humanitarian themes assigned to each report.

- `rw_report_count` – Total number of ReliefWeb reports.
- `rw_food_nutrition_count` – Number of reports classified under Food and Nutrition.
- `rw_agriculture_count` – Number of reports classified under Agriculture.
- `rw_wash_count` – Number of reports classified under Water, Sanitation and Hygiene.
- `rw_health_count` – Number of reports classified under Health.
- `rw_protection_count` – Number of reports classified under Protection and Human Rights.
- `rw_shelter_count` – Number of reports classified under Shelter and Non-Food Items.
- `rw_education_count` – Number of reports classified under Education.

Additional theme-count features can be created for other ReliefWeb themes represented in the study data.

Unlike GDELT, ACLED and ICEWS, the ReliefWeb classification counts represent humanitarian report themes rather than event types. A single ReliefWeb report may contain more than one theme and may therefore contribute to multiple theme-count features.

### 3. Tone Classification Features

Tone classification aims to measure whether the information reported for a particular ADM2 area during a given month is generally more positive or negative. A single tone feature will be created for each source and aggregated to the common **ADM2 × month** unit of analysis.

GDELT already provides an `AvgTone` variable, which measures the average tone associated with the news coverage used to generate an event record. However, an equivalent tone variable is not directly available in ReliefWeb, ACLED or ICEWS.

To create a more consistent measure across the different sources, sentiment analysis using Natural Language Processing (NLP) will be explored. The same sentiment analysis method can be applied to the textual information available from each dataset, such as humanitarian report text, event descriptions and news-related text. Each record will first receive a sentiment score representing how positive or negative the text is.

The individual sentiment scores will then be aggregated by ADM2 area and month to create a single average tone feature for each source. The proposed features are:

- `gdelt_tone_score` – Average tone of GDELT information for an ADM2 area and month.
- `rw_tone_score` – Average tone of ReliefWeb reports for an ADM2 area and month.
- `acled_tone_score` – Average tone of ACLED event descriptions for an ADM2 area and month.
- `icews_tone_score` – Average tone of ICEWS information for an ADM2 area and month.

For GDELT, the original `AvgTone` variable will also be retained as a reference measure. Where suitable news text can be obtained from the original GDELT source information, the NLP-based tone score can be compared with the existing GDELT tone measure.

The purpose of this feature is to provide a standardised measure of reporting tone across the information sources. The tone score represents the positive or negative nature of the reported information rather than a direct measure of food insecurity severity.

### 4. Market Condition Features

Market condition features aim to measure changes in food prices within each ADM2 area over time. Changes in commodity prices can provide information about market conditions and changes in the economic accessibility of food.

The WFP food price dataset contains price observations for multiple food commodities across different markets and administrative areas. Rather than using the raw price of a single commodity, the percentage change in price between consecutive months will be calculated.

For each commodity, the average monthly price within an ADM2 area will first be calculated. The percentage change from the previous month will then be measured using the following formula:

**Price Change (%) = ((Current Month Price - Previous Month Price) / Previous Month Price) × 100**

A positive value indicates that the price of the commodity increased compared with the previous month, while a negative value indicates a decrease. A value close to zero indicates relatively stable prices.

Separate price change features will initially be created for each food commodity available in the dataset. For example:

- `maize_price_change_pct` – Percentage change in maize price compared with the previous month.
- `sorghum_price_change_pct` – Percentage change in sorghum price compared with the previous month.
- `rice_price_change_pct` – Percentage change in rice price compared with the previous month.
- `wheat_price_change_pct` – Percentage change in wheat price compared with the previous month.

The exact commodity features included will depend on the availability of price observations across countries, ADM2 areas and months. Commodities will initially be kept as separate features to preserve the information provided by the WFP dataset and to identify differences in commodity coverage.

After exploring the availability and consistency of the commodity price features, the possibility of combining selected commodities into a broader food price measure may be considered.

### 5. Environmental Condition Features

Environmental condition features aim to capture rainfall conditions within each ADM2 area and month. Rainfall is an important environmental factor associated with agricultural conditions and food availability, particularly in regions where agricultural production and livelihoods depend on seasonal rainfall.

Rainfall information will be obtained from the CHIRPS dataset. Since CHIRPS provides rainfall estimates as spatial raster data, the rainfall values located within each ADM2 boundary will be aggregated to match the common ADM2 × month unit of analysis.

The proposed feature is:

- `rainfall_mean_mm` – Mean monthly rainfall, measured in millimetres, within an ADM2 area.

For each month, the mean rainfall value will be calculated using the CHIRPS rainfall grid cells located within the corresponding ADM2 boundary. This provides a single rainfall value for each ADM2 × month observation.

The purpose of this feature is to provide environmental context alongside the information extracted from the event and humanitarian data sources. The feature represents the amount of rainfall recorded within an area during a given month and does not directly measure drought or food insecurity.
