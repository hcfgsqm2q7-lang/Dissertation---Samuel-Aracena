## Paper Review: *Pushing the Boundaries of Anticipatory Action Using Machine Learning* (Kjærum & Madsen, 2025)

Kjærum and Madsen (2025) present the **Anticipatory Humanitarian Action for Displacement (AHEAD)** model, a machine learning framework developed by the Danish Refugee Council (DRC) to forecast conflict-induced displacement at the subnational level. The objective of the model is to support **anticipatory humanitarian action** by predicting population displacement several months before it occurs, allowing humanitarian organisations to prepare resources and interventions in advance. Unlike traditional humanitarian planning, which often relies on expert judgement and scenario analysis, the AHEAD model combines multiple humanitarian datasets to generate data-driven forecasts that can be incorporated into operational decision-making.

The model is based on a **Bayesian state-space framework**, which estimates the evolution of internally displaced populations over time while incorporating external humanitarian indicators describing conditions within each administrative area and its neighbouring regions. This modelling approach is particularly suitable for humanitarian applications because it can accommodate irregular reporting intervals and missing observations, both of which are common challenges in displacement data. In addition to the state-space model, the authors also developed a national-level forecasting model using an **ensemble of gradient-boosted trees and linear models**, illustrating the growing adoption of machine learning techniques in humanitarian prediction.

Rather than relying on a single source of information, the model integrates features representing multiple humanitarian mechanisms. The features used in the model are summarised below.

| Humanitarian Mechanism | Features | Data Source |
|------------------------|----------|-------------|
| **Conflict and Insecurity** | Violent events, kidnappings, lootings, attacks by extremist groups, fatalities, violence against civilians | ACLED |
| **Food Security** | Percentage of population in **Cadre Harmonisé Phase 3 or above** | Cadre Harmonisé |
| **Market Conditions** | Food prices | World Food Programme (WFP) |
| **Climate and Environmental Stress** | Vegetation Health Index (VHI) | Food and Agriculture Organization (FAO) |
| **Development** | Child mortality rate | World Health Organization (WHO) |
| **Protection** | Community protection indicators (e.g., perceived safety, freedom of movement, access to services) | Project 21 Protection Monitoring System |

One of the main contributions of the paper is the demonstration that humanitarian prediction benefits from integrating information across multiple humanitarian domains rather than relying on a single predictor such as conflict or climate. The authors show that combining conflict, food security, environmental and protection indicators improves the prediction of future displacement compared with simpler baseline approaches. However, they also acknowledge that the model has limitations, particularly in predicting sudden or unprecedented displacement surges, and emphasise that machine learning forecasts should complement, rather than replace, expert judgement and community-based monitoring.

This study is particularly relevant to the present dissertation because it illustrates how heterogeneous humanitarian datasets can be transformed into structured **spatial-temporal features** before being used for predictive modelling. Similar to the proposed **Source-Aware Humanitarian Data Panel**, the AHEAD model aggregates multiple datasets into a common administrative and temporal unit before constructing predictive variables. However, while the AHEAD model focuses on forecasting displacement, the proposed research extends this concept by developing a harmonised humanitarian data panel for food insecurity monitoring that explicitly preserves **source provenance**, documents **data transformations**, and facilitates transparent integration of multiple humanitarian information sources.

## Paper Review: *Forecasting Trends in Food Security with Real-Time Data*

This paper presents a machine learning framework for forecasting short-term food insecurity trends using real-time humanitarian and socio-economic data. The objective of the study is to predict the prevalence of **insufficient food consumption** across multiple countries monitored by the World Food Programme (WFP), allowing humanitarian organisations to identify areas at risk before food insecurity worsens. The authors compare several forecasting approaches, including traditional statistical methods (ARIMA) and multiple machine learning models such as Extreme Gradient Boosting (XGBoost), Long Short-Term Memory (LSTM) networks, Convolutional Neural Networks (CNN), and Reservoir Computing (RC). Their results show that machine learning approaches consistently outperform traditional statistical forecasting methods, with Reservoir Computing providing the best overall performance across the countries analysed.

Rather than relying on a single predictor, the forecasting models integrate information from several humanitarian domains that are known to influence food insecurity. The selected features represent both the current humanitarian situation and the factors that drive changes in food security over time.

| Humanitarian Mechanism | Features Used |
|------------------------|---------------|
| **Historical Food Security** | Previous prevalence of insufficient food consumption, previous reduced Coping Strategy Index (rCSI), Food Consumption Score (FCS) |
| **Conflict and Insecurity** | Number of conflict events, political violence indicators, fatalities (ACLED) |
| **Climate and Environmental Conditions** | Rainfall, Normalised Difference Vegetation Index (NDVI), drought and vegetation indicators |
| **Market Conditions** | Staple food prices, market prices, food price inflation |
| **Population Movement** | Displacement and migration indicators (where available) |
| **Seasonality** | Month of the year and seasonal patterns |

An important aspect of the study is its emphasis on combining multiple humanitarian mechanisms rather than relying on a single source of information. Historical food security indicators are found to be among the strongest predictors of future food insecurity, reflecting the persistent nature of humanitarian crises. However, the authors demonstrate that incorporating conflict, environmental conditions and market information significantly improves forecasting performance, highlighting the importance of integrating complementary humanitarian datasets into predictive models.

The paper is particularly relevant to this dissertation because it illustrates how heterogeneous humanitarian datasets can be transformed into structured features for machine learning applications. Similar to the proposed **Source-Aware Humanitarian Data Panel**, the forecasting framework combines information from multiple humanitarian domains before constructing predictive variables. However, while the primary objective of the paper is to maximise forecasting accuracy, the proposed research focuses on developing a harmonised humanitarian data panel that preserves **source provenance**, documents **feature engineering procedures**, and provides a transparent framework for integrating multiple humanitarian datasets at the **Admin 2 × Month** level. The resulting panel can subsequently be used to support food insecurity prediction and other humanitarian analytical tasks.

## Paper Review: *ViEWS: A Political Violence Early-Warning System* (Hegre et al., 2019)

Hegre et al. (2019) present the **Violence Early-Warning System (ViEWS)**, a machine learning-based forecasting framework designed to predict the occurrence of organised political violence at both national and subnational levels. Unlike traditional conflict research, which primarily focuses on explaining the causes of past conflicts, ViEWS aims to generate operational forecasts that estimate the probability of future political violence several months in advance. The system was developed to support governments, humanitarian organisations and policy makers by providing early warnings that can assist conflict prevention and humanitarian planning.

Rather than relying on a single forecasting model, ViEWS employs an **ensemble forecasting framework** that combines predictions from multiple statistical and machine learning models. Each model captures different theoretical drivers of political violence, and their predictions are combined to produce a final forecast. This ensemble approach improves predictive performance while reducing the uncertainty associated with individual models. Predictions are generated on a **monthly basis** for both country-level and subnational administrative units, making the system suitable for operational early warning applications.

The forecasting framework incorporates features representing multiple dimensions of conflict risk, including historical conflict patterns, political conditions, socioeconomic characteristics, environmental stress and spatial interactions between neighbouring regions.

| Humanitarian Mechanism | Features Used |
|------------------------|---------------|
| **Historical Conflict** | Previous conflict events, previous fatalities, conflict persistence, recent violence |
| **Spatial Diffusion** | Conflict activity in neighbouring administrative areas, regional conflict intensity |
| **Structural Risk** | GDP per capita, demographic characteristics, education, long-term socioeconomic conditions |
| **Political Context** | Political institutions, elections, regime characteristics, political instability |
| **Environmental Conditions** | Rainfall variability, drought and selected climate indicators |
| **Temporal Dynamics** | Lagged conflict variables, previous conflict trends |

One of the principal contributions of ViEWS is its demonstration that conflict forecasting benefits from integrating multiple feature groups rather than relying solely on historical conflict events. The authors show that political violence is influenced by both structural conditions and short-term dynamics, while also exhibiting strong spatial and temporal dependencies. In particular, incorporating information from neighbouring administrative areas significantly improves forecasting performance, highlighting the importance of considering the geographic diffusion of conflict.

### Feature Engineering Strategy

An important aspect of the ViEWS framework is its approach to feature engineering. Instead of using individual conflict events directly, the system aggregates heterogeneous datasets into structured **spatial-temporal indicators** that describe the conditions within each administrative unit for every monthly observation. Historical conflict records, socioeconomic variables, political indicators and environmental conditions are transformed into machine learning features that capture both current conditions and longer-term trends.

The feature engineering process is organised around **theoretical mechanisms of conflict** rather than individual datasets. Variables are grouped according to the processes they represent, such as historical conflict persistence, structural vulnerability, political instability and spatial diffusion. This approach allows the forecasting models to learn relationships between broader humanitarian mechanisms rather than isolated observations.

This methodology closely aligns with the proposed approach of the present dissertation. Similarly, the proposed **Source-Aware Humanitarian Data Panel** aggregates heterogeneous humanitarian datasets into a common **Admin 2 × Month** framework before constructing features representing conflict, climate, market conditions and humanitarian reporting. However, while ViEWS focuses exclusively on forecasting political violence, the proposed research extends this concept to food insecurity monitoring by integrating multiple humanitarian information sources while explicitly preserving **source provenance**, documenting **feature engineering procedures**, and maintaining transparency regarding the origin and uncertainty associated with each feature.

## Paper Review: *Nowcasting Food Insecurity Interest Using Google Trends Data* (Caravaggio et al., 2024)

Caravaggio et al. (2024) investigate whether **Google Trends (GT)** data can be used to monitor food insecurity in near real time. The motivation for the study is that official food security indicators often become available several weeks or months after data collection, limiting their usefulness for early warning systems. The authors propose an alternative approach that leverages online search behaviour to construct a **daily Synthetic Search Interest (SSI)** index representing public interest in food insecurity. Unlike traditional food security monitoring systems that rely on household surveys or humanitarian assessments, the proposed methodology exploits publicly available search engine data to provide more timely indicators of emerging food crises. :contentReference[oaicite:1]{index=1}

Rather than developing a conventional machine learning prediction model, the study combines **Natural Language Processing (NLP)**, **LASSO regression**, and **Google Trends** to identify search terms associated with food insecurity. The methodology begins by analysing **1,414 FEWS NET reports** using NLP to identify the main hazards and shocks associated with food insecurity. LASSO regression is then applied to determine the most important predictors, which are subsequently used to retrieve Google Trends search data across 25 languages. These search volumes are combined into a country-level **Synthetic Search Interest (SSI)** index, weighted according to the importance of each predictor and adjusted for internet penetration to improve comparability across countries. Finally, the SSI index is validated against FAO estimates of the population experiencing severe food insecurity. ## Paper Review: *Natural Language Processing for Humanitarian Action: Opportunities, Challenges, and the Path Toward Humanitarian NLP* (2023)

This paper presents a comprehensive review of how **Natural Language Processing (NLP)** can support humanitarian action by transforming large volumes of unstructured humanitarian text into structured information that can assist decision-making. Rather than proposing a new prediction model, the authors review the current state of humanitarian NLP, discussing existing datasets, machine learning methods, practical applications and remaining challenges. The paper highlights that humanitarian organisations generate vast quantities of textual information through situation reports, needs assessments, news articles, field reports and social media, much of which remains underutilised because it is not available in a structured format.

The review identifies several NLP tasks that have been successfully applied within humanitarian contexts.

| NLP Task | Humanitarian Application |
|----------|--------------------------|
| **Document Classification** | Categorising reports by humanitarian sector (food security, health, protection, shelter, etc.) |
| **Information Extraction** | Identifying locations, organisations, events, affected populations and humanitarian needs from text |
| **Named Entity Recognition (NER)** | Extracting people, places, organisations and other entities from humanitarian documents |
| **Text Summarisation** | Producing concise summaries of lengthy humanitarian situation reports |
| **Question Answering** | Retrieving relevant information from large humanitarian document collections |
| **Machine Translation** | Translating multilingual humanitarian documents into a common language |
| **Sentiment and Text Analysis** | Analysing the tone and content of humanitarian reports to identify emerging trends or concerns |

A central contribution of the paper is its discussion of **humanitarian datasets** developed specifically for NLP research. The authors describe resources such as **DEEP (Data Entry and Exploration Platform)** and **HumSet**, which contain thousands of manually annotated humanitarian documents. These datasets enable supervised machine learning models to classify reports according to humanitarian sectors, disaster types and operational needs. The paper also highlights the increasing use of transformer-based language models, such as **BERT** and its multilingual variants, which have substantially improved performance across a wide range of humanitarian NLP tasks.

### Feature Engineering Strategy

One of the key messages of the paper is that unstructured humanitarian text can be transformed into structured analytical features through NLP. Instead of analysing reports manually, machine learning models can automatically extract information describing humanitarian events, affected populations, geographic locations, response activities and thematic categories. These extracted variables can then be aggregated into structured datasets suitable for statistical analysis and machine learning applications.

The paper also emphasises that humanitarian NLP extends beyond simple text classification. Modern NLP techniques allow researchers to derive quantitative indicators from textual reports, including topic frequencies, document classifications, named entities and semantic representations generated through language models. These structured outputs can then be integrated with traditional humanitarian datasets such as conflict events, climate observations and market indicators.

### Relevance to this Dissertation

This paper is highly relevant to the proposed dissertation because it provides the methodological foundation for incorporating **text-based humanitarian information** into a structured humanitarian data panel. While datasets such as ACLED, CHIRPS and WFP Food Prices already provide structured numerical information, ReliefWeb primarily consists of unstructured humanitarian reports. The NLP techniques described in this paper demonstrate how these reports can be transformed into quantitative features suitable for analysis.

In particular, the review supports the proposed development of a **Tone Classification Feature** for ReliefWeb. Rather than relying solely on manually assigned metadata, NLP methods can be used to analyse the language contained within humanitarian reports and estimate whether the reporting during a particular **Admin 2 × Month** period reflects increasingly positive or negative humanitarian conditions. This approach complements the structured event information extracted from other humanitarian datasets while preserving the source provenance of each feature.

More broadly, the paper reinforces the central idea behind the proposed **Source-Aware Humanitarian Data Panel**: heterogeneous humanitarian information—including structured datasets and unstructured textual reports—can be harmonised into a common spatial-temporal framework to support humanitarian analysis and future predictive modelling.

## Defining the Unit of Analysis

 For this study, the proposed unit of analysis is:

> **One row = one Admin 2 district × one month**

This means that each row in the final dataset will describe the humanitarian conditions in one district during one calendar month.

Admin 2 was selected because it keeps more local detail than national or Admin 1 data. This is useful because neighbouring districts can experience very different levels of conflict, rainfall, food prices or humanitarian activity. Monthly aggregation was chosen because it provides enough temporal detail while still matching the frequency of most of the datasets.

Each source will be transformed to fit this structure. Event datasets such as ACLED, GDELT and ICEWS will be assigned to an Admin 2 district and counted by month. Raster datasets such as CHIRPS and NDVI will be summarised using district-level averages. WFP market prices and ReliefWeb reports will also be linked to districts where location information is available. National indicators such as CPI and exchange rates will be repeated across all districts within the same country and month.

Some datasets may not always provide information at Admin 2 level, so this may create missing values or require the use of broader regional information. However, starting with Admin 2 preserves the highest practical level of detail, and the dataset can always be aggregated to Admin 1 later if needed.

# Defining Mechanisms

Now that we have gone through literature review and understood what other features are being used by other reaserchers, we can now define our set of mechanisms, and map the sources and features relevant to each one.



## 1. Conflict and Insecurity


### Source Dataset(s)

- **ACLED (Armed Conflict Location & Event Data Project)** – provides manually coded records of political violence, protests and strategic developments.
- **GDELT (Global Database of Events, Language and Tone)** – automatically extracts global events from news articles using natural language processing and the CAMEO event coding system.
- **ICEWS (Integrated Crisis Early Warning System)** – contains machine-coded political and conflict events extracted from international news reports using the CAMEO ontology.


### Raw Variables

The following variables are extracted from each dataset:

| Dataset | Raw Variables |
|----------|---------------|
| ACLED | Event date, latitude, longitude, event type, sub-event type |
| GDELT | SQLDATE, EventCode, EventRootCode, QuadClass, ActionGeo location |
| ICEWS | Event date, CAMEO event code, latitude, longitude, source location |

These variables provide the temporal information required to assign events to monthly observations, the spatial information required for geographic aggregation, and the event classifications used to distinguish different forms of conflict.

### Transformation Rule

Conflict events from each dataset are first spatially joined to the selected administrative unit (Admin 1 or Admin 2) using their geographic coordinates. Events are then aggregated by administrative unit and month to produce a common spatial-temporal representation across all datasets.

For each dataset, two groups of features are generated:

- **Total event count:** total number of conflict-related events occurring within each administrative unit during a given month.
- **Event category counts:** number of events belonging to each conflict category (e.g., protests, battles, violence against civilians), based on the event taxonomy provided by each dataset.

Event counts are calculated independently for ACLED, GDELT and ICEWS, preserving the provenance of each source while allowing future comparisons between reporting systems.

### Final Features

The engineered features representing the conflict mechanism include:

| Feature | Description |
|----------|-------------|
| ACLED Event Count | Total number of ACLED events per administrative unit and month |
| ACLED Event Category Counts | Counts of ACLED events by event type |
| GDELT Event Count | Total number of GDELT events per administrative unit and month |
| GDELT Event Category Counts | Counts of GDELT events by event category |
| ICEWS Event Count | Total number of ICEWS events per administrative unit and month |
| ICEWS Event Category Counts | Counts of ICEWS events by event category |

### Expected Relevance to Food Insecurity

Conflict is consistently identified in the literature as one of the strongest predictors of food insecurity. Armed violence can reduce agricultural production, disrupt food supply chains, increase market prices, displace communities and restrict humanitarian access. By capturing both the frequency and type of conflict events, these features aim to quantify the level of insecurity experienced within each administrative unit over time.

### Known Limitations

Several limitations should be considered when interpreting conflict-derived features:

- The three datasets employ different event coding methodologies and therefore may classify similar events differently.
- News-based datasets (GDELT and ICEWS) are subject to reporting bias, media attention and differences in news coverage across regions.
- The same real-world event may appear in multiple datasets, although source provenance is intentionally preserved rather than attempting to remove duplicates.
- Geographic precision varies between datasets, potentially introducing uncertainty during spatial aggregation.

## 2. Climate and Agricultural Stress

### Source Dataset(s)

The climate and agricultural stress mechanism is represented using the **CHIRPS** rainfall dataset.


### Raw Variables

The following variable is extracted from CHIRPS:

| Dataset | Raw Variable |
|----------|--------------|
| CHIRPS | Monthly rainfall (mm) |

Rainfall is provided as a raster dataset, where each grid cell represents the total precipitation accumulated during a given month.

### Transformation Rule

Monthly rainfall rasters are spatially aggregated to the selected administrative unit (Admin 1 or Admin 2) using zonal statistics. For each administrative unit and month, the **mean rainfall** is calculated by averaging the precipitation values of all raster cells contained within the administrative boundary.

The resulting monthly rainfall estimates are then integrated into the common humanitarian panel using the selected spatial-temporal unit of analysis.

### Final Features

The engineered climate feature is:

| Feature | Description |
|----------|-------------|
| Mean Monthly Rainfall | Average monthly rainfall (mm) within each administrative unit |

### Expected Relevance to Food Insecurity

Rainfall is a key determinant of agricultural productivity throughout the Horn of Africa. Reduced precipitation can lead to crop failure, poor pasture conditions, livestock losses and water shortages, while excessive rainfall may cause flooding and damage agricultural production. Consequently, rainfall is widely used as an environmental predictor in food insecurity forecasting models and humanitarian early warning systems.

### Known Limitations

Several limitations should be considered when interpreting rainfall-derived features:

- Rainfall alone does not fully represent agricultural conditions, as crop productivity is also influenced by soil characteristics, irrigation practices and farming techniques.
- Mean rainfall does not distinguish between beneficial rainfall and extreme rainfall events such as floods.
- Monthly averages may mask short periods of intense rainfall or prolonged dry spells occurring within the same month.
- CHIRPS provides precipitation estimates rather than direct observations and may contain uncertainty in areas with limited ground station coverage.

## 3. Market and Food Price Pressure.

### Source Dataset(s)

The market and food price pressure mechanism is represented using the **WFP Food Prices** dataset.

The dataset contains historical observations of food commodity prices collected from markets across multiple countries. Each observation includes information about the commodity being sold, market location, observation date, unit of measurement and reported price. The dataset provides one of the most comprehensive publicly available sources of food price information for humanitarian analysis.

### Raw Variables

The following variables are extracted from the WFP Food Prices dataset:

| Dataset | Raw Variables |
|----------|---------------|
| WFP Food Prices | Observation date, market location, commodity, price, unit of measurement, currency |

These variables provide the temporal information required for monthly aggregation, the spatial information needed to associate markets with administrative regions, and the commodity prices used to quantify market conditions.

### Transformation Rule

Food price observations are first assigned to the selected administrative unit (Admin 1 or Admin 2) according to the location of each market. Prices are then aggregated by administrative unit and month.

Two alternative feature engineering strategies are currently being considered:

**Option 1 – Commodity-specific features**

A small number of essential commodities will be selected for each country based on data availability and their importance within local food markets. Monthly percentage price changes will then be calculated separately for each commodity. This approach preserves the behaviour of individual commodities and allows the analysis to identify which food items experience the greatest price fluctuations.

**Option 2 – Country-specific food basket index**

Multiple essential commodities will be combined to construct a country-specific food basket index representing overall food affordability. Individual commodity prices will first be converted into standardized price indices before being aggregated into a single monthly indicator describing changes in the cost of the food basket.

### Final Features

Depending on the selected feature engineering strategy, the final features will include either:

| Option | Final Feature(s) |
|---------|------------------|
| **Option 1** | Monthly percentage price change for each selected commodity |
| **Option 2** | Food Basket Price Index; Monthly percentage change in Food Basket Price Index |

### Expected Relevance to Food Insecurity

Food prices directly influence household purchasing power and food affordability. Sustained increases in the price of staple commodities reduce access to food, particularly for vulnerable populations that depend on local markets. Previous studies consistently identify food prices as one of the strongest predictors of food insecurity, making market information an essential component of humanitarian early warning systems.

### Known Limitations

Several limitations should be considered when interpreting food price features:

- Market coverage varies between countries and administrative regions.
- Some commodities contain missing observations or inconsistent reporting frequencies.
- Different countries consume different staple foods, making direct comparisons between commodities difficult.
- If a composite food basket index is adopted, assumptions regarding commodity selection and weighting may introduce bias.
- Market prices measure food affordability but do not capture household income or purchasing power directly.

## 4. Economic Conditions

### Source Dataset(s)

The economic conditions mechanism is proposed to be represented using publicly available macroeconomic datasets, including:

- **World Bank Global Economic Monitor (GEM)**
- **International Monetary Fund (IMF)**

These datasets provide monthly economic indicators for a large number of countries and are widely used for macroeconomic analysis and policy monitoring.

### Raw Variables

The following variables are currently being considered:

| Dataset | Raw Variables |
|----------|---------------|
| World Bank GEM / IMF | Consumer Price Index (CPI), Inflation rate |

Additional macroeconomic variables may be explored if they are consistently available for all study countries.

### Transformation Rule

Monthly economic observations will be extracted for each country and aligned with the common temporal framework of the humanitarian panel. Since macroeconomic indicators are generally reported at the national level, the monthly values will be assigned to all administrative units within the corresponding country.

For each variable, monthly percentage changes will be calculated to represent short-term economic fluctuations rather than absolute values. This approach captures changes in economic conditions over time while maintaining consistency with the dynamic features extracted from the other humanitarian datasets.

### Final Features

The proposed engineered features include:

| Feature | Description |
|----------|-------------|
| CPI Monthly Percentage Change | Monthly percentage change in the Consumer Price Index |
| Inflation Rate Monthly Percentage Change | Monthly percentage change in the Inflation rate |

Additional economic indicators may be incorporated if suitable data coverage is identified during the exploratory analysis.

### Expected Relevance to Food Insecurity

Macroeconomic instability can reduce household purchasing power and increase the cost of food through higher inflation and currency depreciation. Rising prices may reduce access to essential goods, particularly in countries that rely on food imports. Consequently, economic indicators provide complementary information to market price data by capturing broader changes in national economic conditions that may influence food insecurity.

### Known Limitations

Several limitations should be considered when interpreting economic indicators:

- The proposed variables are reported at the national level and therefore do not capture economic variation between administrative regions.
- Data availability may differ across countries and time periods, particularly for fragile states.
- Macroeconomic indicators represent broad economic conditions and may not fully reflect local market dynamics.
- Inflation and exchange rates influence food security indirectly and should therefore be interpreted alongside more direct indicators such as food prices and conflict events.

## 5. Humanitarian Access and Response.

### Source Dataset(s)

The humanitarian access and response mechanism is represented using the **ReliefWeb API**.

ReliefWeb aggregates humanitarian reports from governments, United Nations agencies, non-governmental organisations and humanitarian partners worldwide. Each report contains structured metadata describing its publication date, geographic coverage, humanitarian themes, information source and document type, together with the full report text.

### Raw Variables

The following variables are extracted from ReliefWeb:

| Dataset | Raw Variables |
|----------|---------------|
| ReliefWeb | Report ID, publication date, title, report body, country, administrative location (where available), humanitarian themes, report format, source organisation |

These variables provide both structured metadata and unstructured textual information describing humanitarian situations.

### Transformation Rule

Reports are first filtered according to the countries included in the study. Geographic information is then used to assign each report to the selected administrative unit (Admin 1 or Admin 2) whenever sufficient location information is available. Reports are subsequently aggregated by administrative unit and month.

Initially, structured metadata will be used to generate quantitative indicators, including the total number of reports and the number of reports associated with different humanitarian themes. As an extension, Natural Language Processing (NLP) techniques may be applied to the report text to derive additional indicators describing the overall humanitarian context, such as sentiment or humanitarian tone.

### Final Features

The proposed engineered features include:

| Feature | Description |
|----------|-------------|
| Report Count | Total number of ReliefWeb reports per administrative unit and month |
| Report Count by Theme | Number of reports associated with each humanitarian theme |
| Humanitarian Tone Score *(proposed)* | NLP-derived measure describing the overall tone of humanitarian reporting |

Additional text-derived indicators may be explored during later stages of the project.

### Expected Relevance to Food Insecurity

Humanitarian reports often describe crisis conditions before they are fully reflected in structured humanitarian datasets. They provide information on humanitarian operations, food assistance, access constraints, emerging disasters and other contextual factors influencing food security. By combining structured report metadata with information extracted from report text, ReliefWeb provides complementary information that may improve the understanding of humanitarian conditions beyond traditional event or environmental indicators.

### Known Limitations

Several limitations should be considered when interpreting ReliefWeb-derived features:

- Reports describe humanitarian situations rather than direct observations of events.
- Geographic precision varies considerably between reports, with some documents referring to national conditions while others describe specific administrative regions.
- Reporting intensity may reflect organisational priorities, funding availability or media attention rather than the true severity of humanitarian conditions.
- NLP-derived features depend on the quality of text preprocessing and the selected language models, introducing additional methodological uncertainty.
